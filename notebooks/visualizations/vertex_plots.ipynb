```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from sklearn.decomposition import PCA
from scipy.spatial import ConvexHull

# Original 3D points (lying in a 2D affine subspace)
points = np.array([
    [0.3, 0.6, 0.1],
    [0.2, 0.2, 0.6],
    [0.7, 0.1, 0.2],
    [0.3, 0.1, 0.6]
])

# Step 1: Project to 2D using PCA
pca = PCA(n_components=2)
points_2d = pca.fit_transform(points)

# Step 2: Get the correct order using ConvexHull in 2D
hull = ConvexHull(points_2d)
ordered_indices = hull.vertices

# Close the polygon loop
ordered_indices = np.append(ordered_indices, ordered_indices[0])

# Step 3: Plot in 3D using the ordered points
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.view_init(elev=30, azim=45)

# Plot vertices
ax.scatter(points[:, 0], points[:, 1], points[:, 2], color='red')

ax.scatter(samples[:, 0], samples[:, 1], samples[:, 2], color='blue', s=5)

# Construct ordered polygon in 3D
polygon = [points[ordered_indices]]

# Add the polygon
poly = Poly3DCollection(polygon, alpha=0.5, facecolor='lightblue', edgecolor='black')
ax.add_collection3d(poly)

# Axis settings
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_zlim(0, 1)
ax.set_box_aspect([1, 1, 1])
plt.title("10,000 Uniform Samples over Original Polytope in 3D")
plt.savefig("../data/plots/coetzer_reparam_samples_3d.png")
```
