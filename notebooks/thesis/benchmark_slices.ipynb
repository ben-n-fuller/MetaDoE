# Overview
This notebook plots 2D slices of the four chosen benchmark objective functions: Griewank, Ackley, Rastrigin, and Rosenbrock.

```python
import numpy as np
from metadoe.visualization.objectives import griewank, ackley, rastrigin, rosenbrock, plot_benchmark_surface
```

```python
plot_benchmark_surface(
    func=griewank,
    x_range=(-100, 100),
    y_range=(-100, 100),
    dim=3,
    slice_dims=(0, 1),
    fixed_coords=[0.0, 0.0, 0.0],
    n=500,
    title="Griewank (slice at z = 0)",
    save_path="benchmarks/griewank_slice.png",
)
```

```python
plot_benchmark_surface(
    func=ackley,
    x_range=(-5, 5),
    y_range=(-5, 5),
    dim=2,
    slice_dims=(0, 1),
    fixed_coords=[0.0, 0.0],
    n=500,
    title="Ackley",
    save_path="benchmarks/ackley_slice.png",
)
```

```python
plot_benchmark_surface(
    func=rastrigin,
    x_range=(-10, 10),
    y_range=(-10, 10),
    dim=2,
    slice_dims=(0, 1),
    fixed_coords=[0.0, 0.0],
    n=500,
    title="Rastrigin",
    save_path="benchmarks/rastrigin_slice.png",
)
```

```python
plot_benchmark_surface(
    func=rosenbrock,
    x_range=(-20, 20),
    y_range=(-20, 20),
    dim=2,
    slice_dims=(0, 1),
    fixed_coords=[0.0, 0.0],
    n=500,
    title="Rosenbrock",
    save_path="benchmarks/rosenbrock_slice.png",
)
```
