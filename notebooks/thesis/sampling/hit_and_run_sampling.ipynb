# Overview
This notebook is used to generate large samples of constrained regions for visualization.

```python
include("../../../src/MetaDoE.jl")
using .MetaDoE: Experiments, ConstraintEnforcement, Constraints, PSO, Objectives, Designs, Models, HitAndRun
using NPZ
```

## Constrained 3D Region

```python
N = 1
K = 3
bound = 100

A, b = Constraints.hypercube(K, bound)
experiment = Experiments.create(N, K, Models.linear)
experiment = Experiments.with_linear_constraints(experiment, A, b)
for i in 1:(K-1)
    experiment = Experiments.with_factor_ratio(experiment, i, K, 2.0)
end

Constraints.save_vertices(experiment; location = "data/3d_constrained_region.npy")
sampler = HitAndRun.hit_and_run(experiment.constraints.A, experiment.constraints.b)
samples = sampler(10000)
npzwrite("data/3d_constrained_region_samples.npy", samples)
```

```python
import numpy as np
from metadoe.geometry import plot_polyhedron_samples
```

```python
points = np.load("data/3d_constrained_region.npy")
samples = np.load("data/3d_constrained_region_samples.npy")
plot_polyhedron_samples(points, samples, "10k Hit-and-Run Samples over 3D Constrained Region", "plots/3d_constrained_region_with_samples.png", azim=200, elev=15, figsize=(12, 8))
```
